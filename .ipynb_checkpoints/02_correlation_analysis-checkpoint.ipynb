# 02 — Correlation Analysis

**ML-Assisted Electrochemical Descriptor Analysis of Plasmonic Ag/Au g-C3N4 Photocatalysts**

Descriptor-activity correlations at two scales:

- **Section A** — catalyst level (n=7), Pearson correlation per illumination condition
- **Section B** — measurement level (n=84), metal-stratified Pearson correlation on the master table
- **Sections C-I** — supporting figures (flat-band potential, combined LSV, exploratory PLS/RF, chronoamperometry and R_ct heatmaps)

**Inputs:** `Results/TABLE2_catalyst_n28.csv`, `Results/MASTER_TABLE_n84.csv`
**Run after** `01_data_preparation.ipynb`.

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

from utils import SAMPLE_ORDER, SAMPLE_COLORS, SAMPLE_MARKERS, CONDITION_ORDER, CONDITION_COLORS

sample_order = SAMPLE_ORDER
sample_colors = SAMPLE_COLORS
sample_markers = SAMPLE_MARKERS
condition_order = CONDITION_ORDER
cond_colors = CONDITION_COLORS

df7 = pd.read_csv("Results/TABLE2_catalyst_n28.csv")    # n=7 per condition
df84 = pd.read_csv("Results/MASTER_TABLE_n84.csv")       # n=84 full master

print(f"TABLE2 (n=28): {df7.shape}")
print(f"MASTER TABLE (n=84): {df84.shape}")
```

## A. Catalyst-Level Correlation (n=7)

For each illumination condition separately, descriptors are correlated with `J_m0p7`.
At n=7, a two-tailed p<0.05 requires |r| > 0.75.

Dark-condition rows use a reduced descriptor set (Vfb, Eg, Rs, Rct, tau, phase, Cdl) since
the photo-response descriptors (delta_Rct, CA amplitude, CA stability) are undefined without
illumination.

### A1. Correlation Heatmap per Condition

```python
descriptors = [
    "Vfb_V", "Eg_eV", "Rs_ohm", "Rct_ohm", "tau_s",
    "phase_min", "Cdl_est_F", "delta_Rct",
    "CA_dJ_mean_mA", "CA_stability"
]
target = "J_m0p7"
conditions = condition_order

dark_descriptors = [
    "Vfb_V", "Eg_eV", "Rs_ohm", "Rct_ohm", "tau_s", "phase_min", "Cdl_est_F"
]

fig, axes = plt.subplots(1, 4, figsize=(22, 6))
fig.patch.set_facecolor("white")

for ax, cond in zip(axes, conditions):
    feats = dark_descriptors if cond == "Dark" else descriptors
    sub = df7[df7["condition"] == cond][feats + [target]].copy()
    sub = sub.dropna(axis=1, how="all").dropna(axis=0, how="any")

    print(f"{cond}: n={len(sub)}, features={list(sub.columns)}")

    if len(sub) < 3:
        ax.text(0.5, 0.5, f"Insufficient data\n(n={len(sub)})",
                ha="center", va="center", transform=ax.transAxes)
        ax.set_title(f"{cond}\n(n={len(sub)})", fontweight="bold", fontsize=10)
        continue

    corr = sub.corr(method="pearson")
    sns.heatmap(
        corr, ax=ax, cmap="RdBu_r", vmin=-1, vmax=1,
        annot=True, fmt=".2f", annot_kws={"size": 7},
        square=True, linewidths=0.5,
        xticklabels=corr.columns, yticklabels=corr.columns
    )
    ax.set_title(f"{cond}\n(n={len(sub)})", fontweight="bold", fontsize=10)
    ax.tick_params(labelsize=7)

plt.suptitle("Pearson Correlation Heatmap — n=7 per Condition",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("Results/correlation_heatmap_per_condition.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/correlation_heatmap_per_condition.png")
```

**Findings:**

- **Dark** — Vfb is the strongest predictor (r=+0.80, p=0.031): the Schottky barrier controls dark HER activity.
- **410 nm** — tau and Rct emerge as the leading descriptors: charge-transfer kinetics dominate near the Ag LSPR resonance.
- **440 nm** — tau is the top predictor (r=-0.90, p=0.005): faster carriers give the highest J at g-C3N4 band-gap excitation.
- **Full spectrum** — no single dominant descriptor, consistent with broadband excitation activating multiple mechanisms at once.

Descriptor importance shifts with wavelength, consistent with a wavelength-selective mechanism.

### A2. Key Pairwise Scatter Plots

```python
pairs = [
    ("Vfb_V", "J_m0p7", "Dark",   "V$_{fb}$ vs J — Dark (Schottky control)"),
    ("Vfb_V", "J_m0p7", "410 nm", "V$_{fb}$ vs J — 410 nm (mechanism shift)"),
    ("tau_s",  "J_m0p7", "440 nm", "τ vs J — 440 nm (carrier dynamics)"),
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.patch.set_facecolor("white")

for ax, (x_col, y_col, cond, title) in zip(axes, pairs):
    sub = df7[df7["condition"] == cond].copy()
    x = pd.to_numeric(sub[x_col], errors="coerce")
    y = pd.to_numeric(sub[y_col], errors="coerce")
    mask = x.notna() & y.notna()
    x, y = x[mask], y[mask]
    samples = sub["sample"].values[mask]

    for xi, yi, s in zip(x, y, samples):
        ax.scatter(xi, yi, color=sample_colors[s],
                   marker=sample_markers[s], s=120, zorder=5,
                   edgecolors="black", linewidths=0.5)
        ax.annotate(s, (xi, yi), textcoords="offset points",
                    xytext=(5, 3), fontsize=7)

    r, p = stats.pearsonr(x, y)
    m, b = np.polyfit(x, y, 1)
    x_line = np.linspace(x.min(), x.max(), 100)
    ax.plot(x_line, m*x_line+b, color="gray", linewidth=1.5, linestyle="--")
    ax.set_xlabel(x_col, fontsize=10)
    ax.set_ylabel(y_col, fontsize=10)
    ax.set_title(f"{title}\nr={r:+.3f}, p={p:.3f}", fontweight="bold", fontsize=9)
    ax.grid(True, alpha=0.3)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.suptitle("Key Pairwise Correlations — Descriptor vs Photocurrent",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("Results/pairwise_scatter_3.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/pairwise_scatter_3.png")
```

Three key relationships:

- **Vfb vs J, Dark (r=+0.80)** — a more negative Vfb gives a higher |J|: the Schottky barrier drives dark HER.
- **Vfb vs J, 410 nm (r=-0.17)** — the correlation collapses: Vfb is no longer predictive once LSPR excitation is present, indicating the mechanism has shifted.
- **tau vs J, 440 nm (r=-0.90)** — a shorter relaxation time (faster charge transfer) gives a higher |J|: carrier dynamics dominate at band-gap excitation.

The Vfb-to-tau transition from dark to 440 nm is the central mechanistic finding of this section.

### A3. Correlation Table (n=7)

```python
photo_descriptors = descriptors

print("="*65)
print("Pearson Correlation — n=7 per condition")
print("Threshold for p<0.05 at n=7: |r| > 0.75")
print("="*65)

for cond in conditions:
    feats = dark_descriptors if cond == "Dark" else photo_descriptors
    sub = df7[df7["condition"] == cond][feats + [target]].copy()
    sub = sub.dropna(axis=1, how="all").dropna(axis=0, how="any")

    print(f"\n{cond} (n={len(sub)}):")
    if len(sub) < 3:
        print("  Insufficient data — skipped")
        continue

    print(f"  {'Feature':<22} {'r':>8} {'p':>8} {'Sig':>5}")
    print(f"  {'-'*47}")
    for feat in feats:
        if feat not in sub.columns or target not in sub.columns:
            continue
        valid = sub[[feat, target]].dropna()
        if len(valid) < 3:
            continue
        r, p = stats.pearsonr(valid[feat], valid[target])
        sig = "✓" if p < 0.05 else ""
        print(f"  {feat:<22} {r:>8.3f} {p:>8.3f} {sig:>5}")
```

The printed table gives the exact r/p values behind the heatmap and scatter plots above —
Vfb is the only descriptor reaching significance in the dark, and tau is the only one
reaching significance at 440 nm.

## B. Master-Table Correlation (n=84) — Metal-Stratified

At n=84, p<0.05 requires |r| > 0.21; at n=36 per metal, p<0.05 requires |r| > 0.33.
Ag and Au catalysts are analyzed separately here because they respond to different
descriptor sets — pooling them in the n=7 catalyst-level analysis above would mask
these metal-specific relationships.

### B1. Pearson Table — All / Ag / Au

```python
features_84 = [
    "Rs_ohm", "Rct_ohm", "tau_s", "phase_min", "Cdl_est_F",
    "Vfb_V", "Eg_eV", "delta_Rct", "Rct_ratio", "delta_Rs",
    "Rct_dark", "CA_dJ_mean_mA", "CA_stability",
    "TEM_mean_nm", "SSPL_quenching_ratio", "TRPL_tau_A_ns",
    "XRD_metal_tau_nm", "XRD_gCN_tau_nm", "loading_num", "voltage"
]
target = "J_m0p7"

subsets = {
    "All (n=84)":  df84,
    "Ag (n=36)":   df84[df84["metal"] == "Ag"],
    "Au (n=36)":   df84[df84["metal"] == "Au"],
}

print("="*75)
print(f"{'Feature':<25} {'All (n=84)':>14} {'Ag (n=36)':>14} {'Au (n=36)':>14}")
print(f"{'':-<25} {'r / p':>14} {'r / p':>14} {'r / p':>14}")
print("-"*75)

for feat in features_84:
    line = f"{feat:<25}"
    for label, sub in subsets.items():
        data = sub[[feat, target]].dropna()
        if len(data) < 5:
            line += f"{'—':>14}"
            continue
        r, p = stats.pearsonr(data[feat], data[target])
        sig = "✓" if p < 0.05 else ""
        line += f"  {r:+.3f} p={p:.3f}{sig:>1}"
    print(line)

print("="*75)
print("✓ = p < 0.05")
```

**B1. Pearson correlation table — All / Ag / Au (n=84), interpretation**

| Feature | All (n=84) | Ag (n=36) | Au (n=36) | Interpretation |
|---|---|---|---|---|
| Rct_dark | -0.334✓ | -0.248 | -0.529✓ | Higher dark resistance → lower J; Au more sensitive to baseline impedance |
| CA_stability | -0.334✓ | -0.331✓ | -0.202 | Less stable photoresponse → higher transient J; Ag-driven effect |
| Rct_ratio | -0.251✓ | -0.451✓ | -0.114 | Lower light/dark ratio → stronger LSPR → higher J; Ag-specific fingerprint |
| SSPL_quenching | +0.296✓ | +0.477✓ | +0.064 | Higher PL quenching → better charge transfer → higher J; Ag only |
| loading_num | -0.246✓ | -0.460✓ | -0.118 | Lower Ag loading → higher J; no linear effect for Au |
| Eg_eV | +0.226✓ | +0.375✓ | -0.050 | Wider band gap → higher J for Ag; not significant for Au |
| delta_Rct | -0.203 | -0.243 | -0.399✓ | Larger LSPR resistance drop → higher J; Au-specific |
| delta_Rs | -0.103 | -0.127 | -0.383✓ | Ohmic response under illumination; Au-specific only |
| tau_s | +0.090 | +0.338✓ | +0.028 | Longer EIS timescale → higher Ag J; Ag-specific kinetics |
| TEM_mean_nm | +0.057 | +0.487✓ | -0.102 | Larger Ag particles → higher J — metal-support interaction |
| TRPL_tau_A_ns | -0.058 | +0.495✓ | -0.202 | Longer Ag surface lifetime → higher J |
| XRD_metal_tau_nm | -0.203 | -0.492✓ | -0.220 | Smaller Ag crystallites → higher J (better dispersion) |
| XRD_gCN_tau_nm | -0.195 | -0.458✓ | -0.182 | Smaller g-C3N4 domains → higher Ag activity |
| voltage | 0.000 | 0.000 | 0.000 | J_m0p7 is fixed per sample x condition — no voltage dependence, as expected |

**Reading the table:**

- Ag and Au respond to largely different descriptor sets, which is why the two are always analyzed separately from here on.
- Ag activity correlates with both electrochemical descriptors (Rct_ratio, CA_stability) and characterization descriptors (SSPL, TRPL, TEM, XRD) — Ag performance is morphology-dependent.
- Au activity correlates only with electrochemical descriptors (Rct_dark, delta_Rct, delta_Rs); characterization descriptors are not significant for Au.
- Several descriptors flip sign between the two metals.
- `voltage` r=0.000 confirms the data structure is as expected: J repeats 3x per sample x condition.

These are the uncorrected p-values; Benjamini-Hochberg FDR correction is applied on top of
this table in Notebook 05, which removes two of the borderline Ag hits (`CA_stability`,
`CA_dJ_mean_mA`) from the significant set.

### B2. Metal-Stratified Bar Plot

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 8))

subset_list = [
    ("All (n=84)", df84),
    ("Ag only (n=36)", df84[df84["metal"] == "Ag"]),
    ("Au only (n=36)", df84[df84["metal"] == "Au"]),
]

for ax, (label, sub) in zip(axes, subset_list):
    corr_data = sub[features_84 + [target]].dropna()
    corr_matrix = corr_data.corr()[[target]].drop(target)

    colors = ["#E74C3C" if v < 0 else "#2471A3" for v in corr_matrix[target]]
    ax.barh(corr_matrix.index, corr_matrix[target],
            color=colors, alpha=0.8, edgecolor="white")

    for i, feat in enumerate(corr_matrix.index):
        data = sub[[feat, target]].dropna()
        if len(data) < 5:
            continue
        r, p = stats.pearsonr(data[feat], data[target])
        if p < 0.05:
            ax.text(corr_matrix.loc[feat, target] +
                    (0.02 if corr_matrix.loc[feat, target] >= 0 else -0.02),
                    i, "✓", va="center", ha="center",
                    fontsize=10, fontweight="bold", color="black")

    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-1, 1)
    ax.set_title(label, fontweight="bold", fontsize=11)
    ax.set_xlabel("Pearson r vs J_m0p7", fontsize=10)
    ax.grid(True, alpha=0.3, axis="x")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.suptitle("Pearson Correlation vs J_m0p7 — All / Ag / Au\n"
             "Master Table (n=84) | ✓ = p < 0.05",
             fontweight="bold", fontsize=13)
plt.tight_layout()
plt.savefig("Results/pearson_master_metal_split.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/pearson_master_metal_split.png")
```

The bar heights mirror the r-values in the B1 table above; check-marked bars clear the
n=84 (|r|>0.21) or n=36-per-metal (|r|>0.33) significance threshold.

### B3. Metal vs Descriptors Bar Plot

```python
descriptors_plot = [
    ("Rct_ohm",   "Dark",   "R_ct (Ω)",              "Dark R_ct by Metal"),
    ("delta_Rct", "410 nm", "ΔR_ct (Ω)",             "ΔR_ct at 410 nm by Metal"),
    ("J_m0p7",    "410 nm", "J at −0.7 V (mA cm⁻²)", "J at 410 nm by Metal"),
]

x_positions = {
    "C3N4": 0,
    "Ag0.25": 1, "Ag0.5": 2, "Ag1": 3,
    "Au0.25": 4, "Au0.5": 5, "Au1": 6,
}
x_labels = ["C3N4", "Ag\n0.25", "Ag\n0.5", "Ag\n1.0",
             "Au\n0.25", "Au\n0.5", "Au\n1.0"]

fig, axes = plt.subplots(1, 3, figsize=(16, 5.5))
fig.patch.set_facecolor("white")

for ax, (desc, cond, ylabel, title) in zip(axes, descriptors_plot):
    sub = df7[df7["condition"] == cond].copy()
    sub[desc] = pd.to_numeric(sub[desc], errors="coerce")

    for _, row in sub.iterrows():
        sample = row["sample"]
        if sample not in x_positions:
            continue
        x = x_positions[sample]
        val = row[desc]
        if pd.isna(val):
            continue
        if sample == "C3N4":
            color, marker, size = "#555555", "o", 150
        elif "Ag" in sample:
            color, marker, size = "#2471A3", "s", 120
        else:
            color, marker, size = "#E67E22", "^", 120
        ax.scatter(x, val, color=color, marker=marker,
                   s=size, zorder=5, edgecolors="black", linewidths=0.6,
                   label=sample)

    ag_x = [x_positions[f"Ag{l}"] for l in ["0.25", "0.5", "1"]]
    ag_y = [sub[sub["sample"] == f"Ag{l}"][desc].values[0]
             if len(sub[sub["sample"] == f"Ag{l}"]) > 0 else np.nan
             for l in ["0.25", "0.5", "1"]]
    ax.plot(ag_x, ag_y, color="#2471A3", linewidth=1.2, linestyle="--", alpha=0.6)

    au_x = [x_positions[f"Au{l}"] for l in ["0.25", "0.5", "1"]]
    au_y = [sub[sub["sample"] == f"Au{l}"][desc].values[0]
             if len(sub[sub["sample"] == f"Au{l}"]) > 0 else np.nan
             for l in ["0.25", "0.5", "1"]]
    ax.plot(au_x, au_y, color="#E67E22", linewidth=1.2, linestyle="--", alpha=0.6)

    ax.set_xticks(list(range(7)))
    ax.set_xticklabels(x_labels, fontsize=9)
    ax.set_ylabel(ylabel, fontsize=10)
    ax.set_title(title, fontweight="bold", fontsize=10)
    ax.axhline(0, color="gray", linewidth=0.6, linestyle="-", alpha=0.4)
    ax.grid(True, alpha=0.3, axis="y")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.axvline(0.5, color="gray", linestyle=":", alpha=0.4)
    ax.axvline(3.5, color="gray", linestyle=":", alpha=0.4)
    ax.text(0, ax.get_ylim()[1]*0.97, "g-C₃N₄", ha="center", fontsize=8, color="#555555")
    ax.text(2, ax.get_ylim()[1]*0.97, "Ag", ha="center", fontsize=8, color="#2471A3")
    ax.text(5, ax.get_ylim()[1]*0.97, "Au", ha="center", fontsize=8, color="#E67E22")

plt.suptitle("Metal Series Comparison — Key Descriptors",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("Results/metal_vs_descriptors.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/metal_vs_descriptors.png")
```

Dark R_ct decreases with metal loading for both series, consistent with Schottky-junction
strengthening, while delta_Rct and J at 410 nm both peak at intermediate-to-high Ag loading —
consistent with Ag's LSPR resonance sitting at 410 nm.

### B4. Loading vs Descriptors

```python
descriptors_load = [
    ("J_m0p7",    "J at −0.7 V (mA cm⁻²)",  "Photocurrent J by Loading"),
    ("delta_Rct", "ΔR_ct (Ω)",               "LSPR Response ΔR_ct by Loading"),
    ("%enh_m0p7", "% Enhancement at −0.7 V", "% Enhancement by Loading"),
]

loadings = ["0.25", "0.5", "1"]

fig, axes = plt.subplots(1, 3, figsize=(15, 5.5))
fig.patch.set_facecolor("white")

for ax, (desc, ylabel, title) in zip(axes, descriptors_load):
    sub = df7[df7["metal"].isin(["Ag", "Au"])].copy()
    sub[desc] = pd.to_numeric(sub[desc], errors="coerce")
    x = np.arange(len(conditions))
    width = 0.12

    metals = ["Ag", "Au"]
    m_colors = {"Ag": "#2471A3", "Au": "#E67E22"}
    alphas = [0.5, 0.75, 1.0]

    for mi, metal in enumerate(metals):
        for li, (load_str, alpha) in enumerate(zip(loadings, alphas)):
            vals = []
            for cond in conditions:
                row = sub[(sub["metal"] == metal) &
                          (sub["loading"].astype(str) == load_str) &
                          (sub["condition"] == cond)]
                vals.append(row[desc].values[0] if len(row) > 0 else np.nan)
            pos = x + (mi * len(loadings) + li - len(metals)*len(loadings)/2 + 0.5) * width
            ax.bar(pos, vals, width=width,
                   color=m_colors[metal], alpha=alpha, edgecolor="white",
                   label=f"{metal}{load_str}")

    ax.set_xticks(x)
    ax.set_xticklabels(conditions, fontsize=8, rotation=15)
    ax.set_ylabel(ylabel, fontsize=9)
    ax.set_title(title, fontweight="bold", fontsize=9)
    ax.axhline(0, color="black", linewidth=0.5)
    ax.grid(True, alpha=0.3, axis="y")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.suptitle("Loading × Condition Comparison — Ag and Au",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("Results/loading_vs_descriptors.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/loading_vs_descriptors.png")
```

Photocurrent and delta_Rct both trend with loading in a metal-dependent way, confirming
the correlation directions in the B1 table hold up as visible trends across the loading
series, not just as a summary statistic.

## C. Flat-Band Potential by Catalyst

```python
vfb_data = df7.groupby("sample")["Vfb_V"].first().reindex(sample_order).reset_index()

fig, ax = plt.subplots(figsize=(9, 6))
bars = ax.bar(range(len(sample_order)),
              vfb_data["Vfb_V"],
              color=[sample_colors[s] for s in sample_order],
              edgecolor="white", linewidth=0.8, width=0.6, zorder=3)

for bar, val in zip(bars, vfb_data["Vfb_V"]):
    ax.text(bar.get_x() + bar.get_width()/2, val - 0.03,
            f"{val:.3f}", ha="center", va="top",
            fontsize=9, fontweight="bold", color="white")

ax.axhline(0, color="#2ECC71", linewidth=1.5,
           linestyle="--", label="H⁺/H₂ (0 V vs RHE)", zorder=2)

ax.set_xticks(range(len(sample_order)))
ax.set_xticklabels(sample_order, fontsize=11, fontweight="bold")
ax.set_ylabel("Flat-Band Potential $V_{fb}$ (V vs RHE)", fontsize=12)
ax.set_title("Flat-Band Potential per Catalyst\n(Mott-Schottky Analysis)",
             fontsize=13, fontweight="bold")
ax.set_ylim(-1.3, 0.2)
ax.legend(fontsize=10, loc="lower right")
ax.grid(True, alpha=0.3, axis="y", zorder=0)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.text(0, 0.15, "g-C₃N₄", ha="center", fontsize=9, color="#555555")
ax.text(2, 0.15, "Ag series", ha="center", fontsize=9, color="#2471A3")
ax.text(5, 0.15, "Au series", ha="center", fontsize=9, color="#E67E22")

plt.tight_layout()
plt.savefig("Results/Vfb_barplot.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/Vfb_barplot.png")

print("\nVfb values:")
for _, row in vfb_data.iterrows():
    print(f"  {row['sample']:<8} {row['Vfb_V']:+.3f} V vs RHE")
```

Vfb is non-monotonic with loading for both metals (most negative at intermediate Ag
loading, most negative at low Au loading), so Vfb is treated as a per-catalyst property
rather than assumed to scale linearly with loading.

## D. Combined LSV at −0.6 V

```python
pivot = df7.pivot_table(index="sample", columns="condition",
                        values="J_m0p6", aggfunc="first").loc[sample_order]

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(sample_order))
width = 0.2
offsets = [-1.5, -0.5, 0.5, 1.5]

for i, cond in enumerate(conditions):
    vals = [pivot.loc[s, cond] if cond in pivot.columns else 0
            for s in sample_order]
    ax.bar(x + offsets[i]*width, vals,
           width=width, color=cond_colors[cond],
           label=cond, alpha=0.85, edgecolor="white")

ax.set_xticks(x)
ax.set_xticklabels(sample_order, fontsize=11, fontweight="bold")
ax.set_ylabel("J at −0.6 V vs RHE (mA cm⁻²)", fontsize=11)
ax.set_title("Photocurrent Density at −0.6 V vs RHE\nAll Catalysts × All Conditions",
             fontsize=13, fontweight="bold")
ax.legend(title="Condition", fontsize=9, title_fontsize=10, loc="lower left")
ax.axhline(0, color="black", linewidth=0.8)
ax.grid(True, alpha=0.3, axis="y")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.axvline(0.5, color="gray", linestyle="--", alpha=0.4, linewidth=1)
ax.axvline(3.5, color="gray", linestyle="--", alpha=0.4, linewidth=1)

plt.tight_layout()
plt.savefig("Results/LSV_combined_m0p6.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: Results/LSV_combined_m0p6.png")
```

Combined photocurrent density across all four illumination conditions for all seven

## E. Supplementary: In-Sample PLS Regression and RF Importance (n=7)

Reported for completeness only. At n=7 there is no way to cross-validate, so these
results are descriptive/exploratory and are not used to support any of the validated
model conclusions elsewhere in this project (see Notebook 04's SHAP caveat for the same
reasoning applied at n=84).

```python
from sklearn.cross_decomposition import PLSRegression
from sklearn.ensemble import RandomForestRegressor

pls_rf_results = []

for cond in conditions:
    feats = dark_descriptors if cond == "Dark" else photo_descriptors
    sub = df7[df7["condition"] == cond][feats + [target]].copy()
    sub = sub.dropna(axis=1, how="all").dropna(axis=0, how="any")

    if len(sub) < 4:
        print(f"{cond}: insufficient data (n={len(sub)}), skipped")
        continue

    X = sub[[c for c in feats if c in sub.columns]].values
    y = sub[target].values
    feat_names = [c for c in feats if c in sub.columns]

    n_comp = min(2, X.shape[0] - 1, X.shape[1])
    pls = PLSRegression(n_components=n_comp)
    pls.fit(X, y)
    r2_pls = pls.score(X, y)  # in-sample only

    rf = RandomForestRegressor(n_estimators=300, random_state=42)
    rf.fit(X, y)
    importances = dict(zip(feat_names, rf.feature_importances_))
    top3 = sorted(importances.items(), key=lambda x: -x[1])[:3]

    pls_rf_results.append({
        "Condition": cond, "n": len(sub), "n_components": n_comp,
        "PLS_R2_insample": round(r2_pls, 3),
        "RF_top3": top3
    })
    print(f"{cond} (n={len(sub)}): PLS R²(in-sample)={r2_pls:.3f}, RF top3={top3}")

df_pls_rf = pd.DataFrame(pls_rf_results)
df_pls_rf.to_csv("Results/pls_rf_n7_supplementary.csv", index=False)
print("\nSaved: Results/pls_rf_n7_supplementary.csv")
print("\nAll results in-sample only (n=7 too small for cross-validation);")
print("reported as descriptive/exploratory.")
```

In-sample PLS R² ranges from 0.56 (full spectrum) to 0.90 (dark), with the 440 nm
condition (n=5) reaching R²=0.999 — a textbook overfitting signature at this sample
size rather than genuine predictive signal. RF's top-ranked feature does not consistently
match the validated Pearson findings above, which is expected given RF importance
instability at very small n.

## F. Reconstructing Missing CA_dJ_mean Values

Two catalyst x condition cells (0.25% Ag/g-C3N4 and 0.25% Au/g-C3N4 at 440 nm) are
missing the multi-cycle `CA_dJ_mean_mA` descriptor, even though the underlying
dark/photo current summary values exist for those cells. Where recoverable, the mean
amplitude is reconstructed as `J_photo - J_dark`. `CA_stability` cannot be recovered
this way (it needs the individual first/last-cycle values) and is left missing.

```python
mask_fix = df7["CA_dJ_mean_mA"].isna() & df7["CA_J_photo_mA"].notna() & df7["CA_J_dark_mA"].notna()
print(f"Rows reconstructed: {mask_fix.sum()}")
print(df7[mask_fix][["sample", "condition", "CA_J_dark_mA", "CA_J_photo_mA"]])

df7.loc[mask_fix, "CA_dJ_mean_mA"] = df7.loc[mask_fix, "CA_J_photo_mA"] - df7.loc[mask_fix, "CA_J_dark_mA"]

print("\nAfter reconstruction:")
print(df7[df7["sample"].isin(["Ag0.25", "Au0.25"]) & (df7["condition"] == "440 nm")]
      [["sample", "condition", "CA_dJ_mean_mA", "CA_stability"]])
```

`CA_dJ_mean_mA` is now populated for both previously-missing cells; `CA_stability`
remains genuinely unavailable for these two and is reported as missing rather than
estimated, since there is no physically defensible way to reconstruct a final/initial
cycle ratio from only the summary dark/photo values.
